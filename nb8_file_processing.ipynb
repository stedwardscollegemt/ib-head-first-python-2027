### Python File Objects

In Python, access to files is provided through a *file object*, which is created by calling the built-in `open` function as we have done so before. 

```python
with open(filepath) as file:
        lines = file.readlines()
```

Once the object is created, several functions become available for accessing the file, both for **reading** and **writing**. Aside from the *file path*, the function can also accept a *mode* to indicate why you would like to open the file.

```python
# open syntax
open(filepath, mode)
```

#### Different File Modes

| Mode     | Syntax                          | Description                                |
|----------|---------------------------------|--------------------------------------------|
| `r`      | `open("my_file.txt", "r")`      | Read or error if file does not exist       |
| `w`      | `open("my_file.txt", "w")`      | Write, creates the file if it does not exist|
| `a`      | `open("my_file.txt", "a")`      | Append, creates the file if it does not exist|

Example:
```python
with open("data/my_file.txt", "w") as my_file:
    my_file.write("Hello, my file.")
```

````python
# Example 1: Writing to a new file

with open("data/my_file.txt", "w") as my_file:
    my_file.write("Hello, my file.")
````

#### Write to files, including appending data

As you probably noticed, write mode (`w`) *overwrites* the file if it exists or creates a new one if it doesn't. 

When you want to add new data to an existing file, you should open it in append mode (`a`). Data will be added to the end of the file without losing existing content.

````python
# TODO: Try opening my_file.txt in append mode and add new content to the file
with open("data/my_file.txt", "a") as my_file:
    my_file.write("Adding new data.");
````

### Activity — Student Data Form

Complete the program so that user data is written to the file in the following format:

```
Abi, 10
Ali, 12
Alison, 14
Aurora, 13
Bill, 18
```

````python
with open("data/students_file.txt", "a") as students_file:
    
    entering_data = True
    while entering_data:
        
        name = input("Enter student name: ")
        age = input("Enter student age: ")

        response = input("Would you like to save this data? (y/n): ")
        if response == "y":
            print("Writing your data to the file...")
            
            students_file.write(f"{name},{age}\n") # \n means write a new-line character

        response = input("Would you like to continue entering data? (y/n): ")
        if response == "n":
            entering_data = False
````

### Reading Functions

Python may encounter a problem when opening a file for reading. If it does, it will issue an `IOError` exception, along with a more detailed message describing the problem. 

It is good practice to `open` files inside a `try/except` clause to prevent the program from crashing when there is a problem opening a file.

```python
try:
    with open('my_file.txt', 'r') as my_file:
        # use the file object as needed
except IOError:
    print("We could not open the file.")
```

Once a file has been opened successfully, several methods can be used to access its contents:

`read()` → returns the entire file content as a single string
`readline()` → returns one line from the file
`readlines()` → returns all lines as a list

````python
# Example 1 - readlines()
try:
    with open('data/students_file.txt', 'r') as students_file:
        lines = students_file.readlines()
        print(lines)
except IOError:
    print("Error. We could not open the file.")
````

````python
# Example 2 - read()
try:
    with open('data/students_file.txt') as students_file:
        content = students_file.read()
        print(content)
except IOError:
    print("Error. We could not open the file.")
````
