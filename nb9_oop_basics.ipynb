### Evaluate the fundamentals of OOP

Many textbooks and educational organisations introduce a concept by looking at its advantages and disadvantages. OOP is no exception. However, from a developer's point of view, the more important question is often: *When is OOP a suitable approach, and how can we use it effectively?*

OOP is a widely used programming paradigm, particularly when developing applications that model *real-world entities and their behaviour*. Consider the following examples of popular programming frameworks and libraries:

- Django is a Python web framework that supports object-oriented design, including class-based views and models.
- PyTorch uses classes extensively to represent neural networks and other components used in machine learning.
- Ruby on Rails is a web application framework that makes extensive use of object-oriented design
- Spring is a Java application framework that uses object-oriented concepts and dependency injection.
- Laravel is a PHP framework that uses object-oriented design to build web applications and APIs.

These examples demonstrate how widely OOP is used in software development. However, the popularity of OOP does not mean that it is always the best approach. Procedural and functional programming can also be effective, depending on the problem being solved.

So, how do we evaluate OOP?

OOP is particularly useful when a problem can be modelled as a collection of objects that combine data and behaviour. For example, a school administration system might contain Student, Teacher and Classroom objects, each with its own attributes and methods. This approach can help organise a complex program into manageable components.

However, OOP can also introduce unnecessary complexity. A programmer who creates too many classes, or designs complicated relationships between them, can make a program harder to understand and maintain. Sometimes, a simple procedural solution is all that is needed.

In other words, OOP is a powerful tool, but its effectiveness depends on how well it is designed and whether it is appropriate for the problem.

##### A note on good OOP design: SOLID principles

In professional software development, the SOLID principles are a set of guidelines that can help developers design object-oriented systems that are easier to maintain and extend. Although these principles go beyond the scope of IB Computer Science, two of them are particularly useful to keep in mind as you begin learning OOP.

- **S — Single Responsibility Principle (SRP)**: A class should have one clearly defined responsibility, meaning it should have only one reason to change.

- **L — Liskov Substitution Principle (LSP)**: An instance of a subclass should be usable wherever an instance of its parent class is expected, without breaking the program's expected behaviour.

Applying these principles appropriately can contribute to three important qualities of software:

- Maintainability: Code is easier to understand, correct and update.
- Flexibility: The program can be extended more easily as requirements change.
- Testability: Components with clearly defined responsibilities can be easier to test independently.

There is a trade-off, however. Designing a system around these principles takes practice, and introducing additional classes and relationships can increase the amount of code and the initial complexity of a solution.

> The goal is not to make every program as complex as possible. It is to design a solution that is clear, appropriate for the problem, and easy to maintain as it grows.

### UML to design SOLID OOP

#### What is UML?


**Unified Modelling Language (UML)** is a standard visual way of representing the architecture of a software system that makes use of the OOP paradigm. It is not dependent on knowing a single programming language. It concisely and accurately illustrates the relationships and dependencies between modules of a system. The best way of learning SOLID OOP is by interpreting well-designed UML class diagrams. From a UML, it is then quite straightforward to translate it to object-oriented code.

> For IB, we will focus on UML class diagrams. These represent the structure of classes, including their attributes, methods and relationships.
