When designing and implementing solutions, we tend to analyse **real-world scenarios or objects** of varying complexity. Here is a short list of relatable examples:

### Evaluate the fundamentals of OOP

Many text books and educational organizations think of deconstructing a concept in terms of *pros and cons* — OOP is no exception. However, from a developer's point of view, it rarely works that way. When you frame a concept by weighing advantages alongside the disadvantages, it normally means you are making a choice. But OOP is not a clear choice that you can easily back away from. If you want to **build generic solutions quickly for real-world business scenarios using the best programming frameworks out there**, then you are going to use the OOP paradigm. Here are some examples:

- Python's Django was not intended to be OOP and does not implement it in its purest form, but once you learn the discipline of OOP, the framework is so much easier to learn. 
- Pytorch by Facebook is a framework that allows developers to leverage high performance hardware for the purpose of deep learning and it is designed around OOP principles.
- Ruby on Rails used by X is OOP in its purest form to build high performance web applications.
- Java Spring — OOP is synonymous with Java being the founder programming language of the paradigm.
- PHP Laravel is a very neat framework to build APIs using the OOP paradigm. 

So, I will put it this way for you.

>There is a good way to use OOP i.e., using **SOLID** principles gives you *a lot of advantages*, and a bad way of using OOP i.e., becomes too complex too quickly and hence very difficult to maintain especially when you are part of a development team.

When using OOP, apply the following SOLID principles to gain all of the advantages:

- **S is for Single Responsibility Principle**: A class should have only one job, meaning it should only have one reason to change.
- **O is for Open/Closed Principle (OCP)**: A class should be open for extension, but closed for modification. Always add new behavior by writing *new and independent* code rather than altering existing, tested code.
- **L is for Liskov Substitution Principle (LSP)**: Classes should be substitutable for other related classes without breaking the program.
- **I is for Interface Segregation Principle (ISP)**: An object should not be forced to implement methods it does not need. It is better to write several small, specific classess than one large, general-purpose class.
- **D is for Dependency Inversion Principle (DIP)**: A module should depend on a generic class that is versatile, not on a class that is used for a specific reason.

Applying SOLID principles gives the following key benefits

- **Maintainability**: Code is readable, neat and so easier to debug.
- **Flexibility**: New code can be added with minimal risk of breaking existing code.
- **Testability**: Smaller, independent code is easier to test in isolation.

> But, this way of thinking has a steep learning curve and the amount of code you must write to abide by all of the priciples will increase significantly.

### UML to design SOLID OOP

#### What is UML?

**Unified Modelling Language (UML)** is a standard visual way of representing the architecture of a software system that makes use of the OOP paradigm. It is not dependant on knowing a single programming language. It concisely and accurately illustrates the relationships and dependencies between modules of a system. The best way of learning SOLID OOP is by interpreting *well-designed* UML diagrams. From a UML, it is then very easy to translate it to object-oriented code. Writing SOLID code cold turkey is very difficult to do and it is not recommended.

If you are in a programming team that all know UML then you could use this as a tool for technical and architecture discussions (although from my experience very few are proficient since it is so vast).

A UML diagram is simply a drawing that demonstrates a template for a particular object — a model of a real-world entity that we intend to use in our programs.
